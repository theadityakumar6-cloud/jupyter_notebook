```python
#return the last integer of index xpresent in the array(use slice)

def lastindex(arr, x):
    if len(arr) == 0:
        return -1
    
    if arr[len(arr)-1] == x:
        return len(arr) - 1
        
    smalloutput = lastindex(arr[:len(arr)-1], x)

    if smalloutput == -1:
        return -1
    return smalloutput
    


a = [1,3,7,8,5,7,7,5,4,3,6,7,5,6]
x = 7
print(lastindex(a, x))
```
