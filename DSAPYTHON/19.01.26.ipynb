```python
# Program to check Armstrong number

num = int(input("Enter a number: "))
temp = num
sum_of_cubes = 0

while temp > 0:
    digit = temp % 10
    sum_of_cubes += digit ** 3
    temp = temp // 10

if sum_of_cubes == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
```

```python
# Initial values
num1 = 4
num2 = 3
num3 = 2

# Step 1: num1 += num2 + num3
num1 += num2 + num3
print("After num1 += num2 + num3:", num1)  # 9

# Step 2: num1 = num1 ** (num2 + num3)
num1 = num1 ** (num2 + num3)
print("After num1 = num1 ** (num2 + num3):", num1)  # 59049

# Step 3: Safe version of num1 **= num2 + c
# We define c = 1 to avoid error (example)
c = 1
num1 **= (num2 + c)
print("After num1 **= num2 + c:", num1)  

# Step 4: String concatenation
num1 = '5' + '5'
print("After '5' + '5':", num1)  # 55

# Step 5: Division
print("4.00 / (2.0 + 2.0):", 4.00 / (2.0 + 2.0))  # 1.0

# Step 6: Arithmetic expression
num1 = 2 + 9 * ((3*12) - 8) / 10
print("2 + 9*((3*12)-8)/10:", num1)  # 27.2

# Step 7: Convert int to float
num1 = float(10)
print("float(10):", num1)  # 10.0

# Step 8: Convert string to float then to int safely
num1 = int(float('3.14'))  # Convert to float first, then to int
print("int(float('3.14')):", num1)  # 3
```
