```python
# Program to find GCD/HCF of two numbers

num1 = int(input("Enter first positive number: "))
num2 = int(input("Enter second positive number: "))

# Euclidean algorithm
a, b = num1, num2
while b != 0:
    a, b = b, a % b

print(f"The GCD/HCF of {num1} and {num2} is: {a}")
```

```python
# Program to find LCM of two numbers

num1 = int(input("Enter first positive number: "))
num2 = int(input("Enter second positive number: "))

# LCM formula: LCM * GCD = num1 * num2
a, b = num1, num2
gcd = a
temp = b
while temp != 0:
    gcd, temp = temp, gcd % temp

lcm = (num1 * num2) // gcd
print(f"The LCM of {num1} and {num2} is: {lcm}")
```
